```python
from elasticsearch import Elasticsearch

# Connect to Elasticsearch
es = Elasticsearch("http://localhost:9200")

# Define the index and query
index_name = "triples"
query = {
    "query": {
        "term": {
            "doc_id": doc_id
        }
    },
    'size':100
}

# Execute the search
response = es.search(index=index_name, body=query)

# Print the results
triples = []
for hit in response['hits']['hits']:
    doc = hit['_source']
    triples.append(doc)
```

```python
def get_triples_score(text, triples, model):
    sentences = []
    for triple in triples:        
        sentences.append(f"{triple['head']} {persian_property_set[triple['relation']]} {triple['tail']}")

    sentences_1 = [text]

    embeddings_1 = model.encode(sentences_1,                             
                                max_length=1000, # If you don't need such a long length, you can set a smaller value to speed up the encoding process.
                                )
    embeddings_2 = model.encode(sentences)
    similarity = embeddings_1 @ embeddings_2.T
    
    for i in range(len(triples)):
        triples[i]['score'] = similarity[0][i]
    
    
```

```python
def filter_triples(triples):
  new_triples = []
  for triple in triples:
    if triple['relation'] in persian_property_set.keys():
      new_triples.append(triple)
  return new_triples
filtered_triples = filter_triples(triples)
```

```python
def print_relations(triples):
    for triple in triples:                
        print(f"{triple['head']} {persian_property_set[triple['relation']]} ({property_description[triple['relation']]}) {triple['tail']}")
        print(f"{triple['head_id']} {triple['relation']} {triple['tail_id']}")
        print("="*30)
```

```python
from collections import defaultdict
import numpy as np
from typing import List, Dict

def filter_by_score(data: List[Dict], node) -> List[Dict]:
    """
    Groups the input list of dictionaries by 'head', then by 'head_id',
    and retains only the subgroup (per head) with the highest cumulative score.

    Args:
        data (List[Dict]): List of dictionaries with 'head', 'head_id', and 'score'.

    Returns:
        List[Dict]: Filtered list with only the highest-scoring subgroup per 'head'.
    """
    result = []
    head_groups = defaultdict(list)

    # Group by 'head'
    for item in data:
        head_groups[item[node]].append(item)

    # Within each head group, group by 'head_id' and select highest scoring group
    for head, items in head_groups.items():
        head_id_groups = defaultdict(list)
        for item in items:
            head_id_groups[item[f'{node}_id']].append(item)

        # Select the head_id group with the maximum total score
        max_head_id = max(
            head_id_groups.items(),
            key=lambda x: sum(i['score'] for i in x[1])
        )[0]

        result.extend(head_id_groups[max_head_id])

    return result
```

```python
from groq import Groq

# Your prompt template
prompt_template = """
من یک متن و یک رابطه به تو می‌دهم. بررسی کن که آیا این رابطه در متن ذکر شده یا از محتوای متن قابل برداشت است یا نه. 
فقط با یک کلمه پاسخ بده: «بله» اگر رابطه در متن وجود دارد یا به وضوح از آن استنباط می‌شود، و «خیر» اگر چنین رابطه‌ای در متن وجود ندارد یا قابل استنباط نیست. 
هیچ توضیح اضافه‌ای نده.

متن:
{text}

رابطه:
{head} -------- {relation} -------- {tail}
(توضیح نوع رابطه: {description})

فقط بررسی کن که آیا این رابطه بین این دو موجودیت در متن بالا وجود دارد یا نه.

"""

# Initialize client
client = Groq(api_key='API_KEY') 

# Loop through each set of variables
for variables in variables_list:
    # Format prompt
    formatted_prompt = prompt_template.format(**variables)

    # Make API call
    completion = client.chat.completions.create(
        model="gemma2-9b-it",
        messages=[
            {
                "role": "user",
                "content": formatted_prompt
            }
        ],
        temperature=1,
        max_completion_tokens=1024,
        top_p=1,
        stream=True,
        stop=None,
    )

    print(f"---{variables['head']}---{variables['relation']}---{variables['tail']}")
    print(f"---{variables['head_id']}---{variables['relation_id']}---{variables['tail_id']}")
    for chunk in completion:
        print(chunk.choices[0].delta.content or "", end="")
    print("\n")  # Newline for spacing between outputs
```
